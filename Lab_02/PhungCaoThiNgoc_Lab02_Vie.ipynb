```python
from collections import deque
import heapq
from time import perf_counter
```

# 1. Breadth-First Search

## 1.1 BFS Algorithm

```python
def bfs_shortest_path(graph, start, goal):
    """
    Tìm đường đi ngắn nhất trong đồ thị không trọng số bằng BFS.
    """
    queue = deque([[start]])  # Hàng đợi lưu các đường đi, bắt đầu với ['A']
    visited = set() # Tập đánh dấu các node đã thăm

    while queue:
        path = queue.popleft() # Lấy đường đi đầu tiên ra khỏi hàng đợi
        node = path[-1] # Lấy node cuối cùng trong đường đi hiện tại
        if node in visited: # Bỏ qua nếu node đã được thăm
            continue
        visited.add(node) # Đánh dấu node đã thăm
        if node == goal:
            return path # Nếu node là mục tiêu → trả về đường đi
        # Mở rộng các láng giềng
        for neighbor in graph.get(node, []):
            new_path = list(path)
            new_path.append(neighbor)
            queue.append(new_path) # Tạo đường đi mới và đưa vào hàng đợi

    return None # Không tìm thấy đường đi
```

## 1.1 Input of BFS

```python
# Đồ thị ví dụ
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["F"],
    "F": []
}
start_node = "A"
goal_node = "F"
```

## 1.3 Output of BFS

```python
print(f"Đường đi ngắn nhất {start_node} đến {goal_node}")
print(f"{bfs_shortest_path(graph, start_node, goal_node)}")
```

# 2. Depth-first search

## 2.1 DFS Algorithm 

```python
visited = set() # Tập chứa các node đã thăm
traversal_order = [] # Lưu thứ tự duyệt
def dfs(node):
    if node in visited: # Nếu node đã thăm thì bỏ qua
        return
    
    visited.add(node) # Đánh dấu là đã thăm
    traversal_order.append(node) # Lưu node vào thứ tự duyệt
    
    for neighbor in graph[node]: # Duyệt toàn bộ các hàng xóm chưa thăm
        dfs(neighbor)  # Gọi đệ quy để đi sâu xuống nhánh
```

## 2.2 Input of DFS

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["F"],
    "F": []
}
start_node = "A"
```

## 2.3 Output of DFS

```python
dfs(start_node) # Chạy DFS từ node A
print("Thứ tự duyệt DFS:", traversal_order)
```

# 3. A*

## 3.1 A* Algorithm

```python
# Thuật toán A* để tìm đường đi ngắn nhất trong đồ thị có trọng số
def a_star(graph, heuristic, start, goal):
    pq = [] # Hàng đợi ưu tiên (min-heap), lưu: (f(n), node, path, g(n))
    
    # Đưa node bắt đầu vào hàng đợi
    heapq.heappush(pq, (heuristic[start], start, [start], 0))
    # f(start) = g(start)=0 + h(start)

    visited = set()  # Tập lưu các node đã duyệt

    while pq:
        f, node, path, g = heapq.heappop(pq)
        if node in visited: # Nếu node đã duyệt thì bỏ qua
            continue
        visited.add(node)
        if node == goal: # Nếu đến đích → trả về đường đi
            return path

        # Duyệt hàng xóm của node
        for neighbor, cost in graph.get(node, []):
            if neighbor not in visited:
                new_g = g + cost # chi phí thực tế g(n)
                new_f = new_g + heuristic[neighbor] # f(n) = g(n) + h(n)
                new_path = path + [neighbor]

                # Đưa vào hàng đợi ưu tiên
                heapq.heappush(pq, (new_f, neighbor, new_path, new_g))

    return None  # Không tìm thấy đường đi
```

## 3.2 Input of A*

```python
# Đồ thị có trọng số
graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("D", 2), ("E", 5)],
    "C": [("F", 3)],
    "D": [("F", 1)],
    "E": [("F", 2)],
    "F": []
}

# Heuristic (h(n))
heuristic = {"A": 7, "B": 6, "C": 2, "D": 1, "E": 2, "F": 0}

start_node = "A"
goal_node = "F"
```

## 3.3 Output of A*

```python
print("Đường đi ngắn nhất (A*):")
print(a_star(graph, heuristic, start_node, goal_node))
```

# 4. Compare BFS and A*

## 4.1 Caculate time of BFS and A* 

```python
def compare_bfs_a_star(graph_for_bfs, graph_for_astar, start, goal):
    # ---- BFS ----
    t1 = perf_counter()
    bfs_path = bfs_shortest_path(graph_for_bfs, start, goal)
    t2 = perf_counter()
    bfs_time = t2 - t1

    # ---- A* ----
    t3 = perf_counter()
    a_star_path = a_star(graph_for_astar, heuristic, start, goal)
    t4 = perf_counter()
    a_star_time = t4 - t3

    print("Đường đi BFS :", bfs_path)
    print("Thời gian BFS:", bfs_time)
    print("Đường đi A*  :", a_star_path)
    print("Thời gian A* :", a_star_time)
```

## 4.2 Input of BFS and A*

```python
graph_uniform = {
    'A': {'B': 1, 'C': 1},
    'B': {'D': 1},
    'C': {'D': 1, 'E': 1},
    'D': {},
    'E': {'F': 1, 'G': 1},
    'F': {},
    'G': {}
}
heuristic = {
    'A': 4,
    'B': 3,
    'C': 3,
    'D': 2,
    'E': 2,
    'F': 0,
    'G': 1
}
start_node = "A"
goal_node = "F"

# 1) BFS của bạn dùng dạng adjacency list: node -> [neighbors]
graph_for_bfs = {
    node: list(neighbors.keys())
    for node, neighbors in graph_uniform.items()
}

# 2) A* của bạn dùng dạng weighted adjacency list: node -> [(neighbor, cost), ...]
graph_for_astar = {
    node: [(nbr, cost) for nbr, cost in neighbors.items()]
    for node, neighbors in graph_uniform.items()
}
```

## 4.3 Output of Caculation between BFS and A*

```python
compare_bfs_a_star(graph_for_bfs, graph_for_astar, start_node, goal_node)
```

BFS và A* đều tìm được cùng một đường đi ngắn nhất:
A → C → E → F, đúng như yêu cầu của đồ thị và heuristic.

BFS mất nhiều thời gian hơn vì phải duyệt theo từng tầng và khám phá nhiều node không cần thiết. BFS không sử dụng bất kỳ thông tin định hướng nào, nên bản chất là một thuật toán “mù”.

A* chạy nhanh hơn nhờ sử dụng heuristic để ưu tiên các đường đi có triển vọng, từ đó giảm số node mở rộng. A* vừa đảm bảo tìm đúng đường đi tối ưu, vừa cải thiện hiệu suất đáng kể nếu heuristic là admissible và consistent (đúng như đề bài yêu cầu).
